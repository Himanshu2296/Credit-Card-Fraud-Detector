```python
from pathlib import Path

dataset_path = Path("data/creditcard.csv")

print("Dataset exists: ", dataset_path.exists())
```

```python
import pandas as pd
import numpy as np

print("Credit Card Fraud Detector")
print("Environment ready!")
```

```python
df = pd.read_csv("data/creditcard.csv")

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)

print("Sample of data:\n",df.sample(5))
```

```python
print("Sample data from the dataset\n:",df.sample(5))

print("Info of data\n:",df.info())

print("Total missing values:\n", df.isnull().sum())

class_counts = df["Class"].value_counts().sort_index()

print("Transaction counts:\n")
print("Genuine transactions (Class 0):", class_counts.get(0, 0))
print("Fraudulent transactions (Class 1):", class_counts.get(1, 0))

```

```python
X = df.drop("Class", axis=1)
y = df["Class"]

print("Features shape:", X.shape)
print("Target shape:", y.shape)
```

```python
from sklearn.model_selection import train_test_split

X_temp, X_test, y_temp, y_test = train_test_split(
    X, y,
    test_size=0.15,
    random_state=42,
    stratify=y
)

X_train, X_val, y_train, y_val = train_test_split(
    X_temp, y_temp,
    test_size=0.15 / 0.85,
    random_state=42,
    stratify=y_temp
)

print("Training set:", X_train.shape, y_train.shape)
print("Validation set:", X_val.shape, y_val.shape)
print("Test set:", X_test.shape, y_test.shape)

print("\nFraud cases in training set:", y_train.sum())
print("Fraud cases in validation set:", y_val.sum())
print("Fraud cases in test set:", y_test.sum())
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report

model = LogisticRegression(
    class_weight="balanced",
    max_iter=1000,
    random_state=42
)

model.fit(X_train, y_train)

print("Model training completed!")
```

```python
y_val_pred = model.predict(X_val)

print(classification_report(
    y_val,
    y_val_pred,
    target_names=["Legitimate", "Fraud"],
    zero_division=0
))
```

```python
from sklearn.metrics import confusion_matrix

cm = confusion_matrix(y_val, y_val_pred)

print("Confusion Matrix:")
print(cm)

tn, fp, fn, tp = cm.ravel()

print("\nTrue Negatives:", tn)
print("False Positives:", fp)
print("False Negatives:", fn)
print("True Positives:", tp)
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    average_precision_score,
    precision_score,
    recall_score,
    f1_score
)

rf_model = RandomForestClassifier(
    n_estimators=100,
    class_weight="balanced",
    random_state=42,
    n_jobs=-1
)

rf_model.fit(X_train, y_train)

print("Random Forest training completed!")
```

```python
lr_probs = model.predict_proba(X_val)[:, 1]
rf_probs = rf_model.predict_proba(X_val)[:, 1]

lr_preds = (lr_probs >= 0.5).astype(int)
rf_preds = (rf_probs >= 0.5).astype(int)

def evaluate_model(name, y_true, probabilities, predictions):
    print(f"\n{name}")
    print(f"Average Precision: {average_precision_score(y_true, probabilities):.4f}")
    print(f"Fraud Precision:   {precision_score(y_true, predictions, zero_division=0):.4f}")
    print(f"Fraud Recall:      {recall_score(y_true, predictions, zero_division=0):.4f}")
    print(f"Fraud F1-score:    {f1_score(y_true, predictions, zero_division=0):.4f}")

evaluate_model("Logistic Regression", y_val, lr_probs, lr_preds)
evaluate_model("Random Forest", y_val, rf_probs, rf_preds)
```

```python
X_train_final = pd.concat([X_train, X_val])
y_train_final = pd.concat([y_train, y_val])

final_model = RandomForestClassifier(
    n_estimators=100,
    class_weight="balanced",
    random_state=42,
    n_jobs=-1
)

final_model.fit(X_train_final, y_train_final)

print("Final model training completed!")
```

```python
y_test_pred = final_model.predict(X_test)

y_test_probs = final_model.predict_proba(X_test)[:, 1]

print("Final Test Classification Report:")
print(classification_report(
    y_test,
    y_test_pred,
    target_names=["Legitimate", "Fraud"],
    zero_division=0
))

print("Test Average Precision:",round(average_precision_score(y_test, y_test_probs), 4))

print("Test Confusion Matrix:")
print(confusion_matrix(y_test, y_test_pred))
```

```python
import joblib

joblib.dump(final_model, "fraud_model.joblib")

print("Model saved successfully!")
```

```python
loaded_model = joblib.load("fraud_model.joblib")

row_number = int(input("Enter test transaction row number (0 to 42720): "))

if 0 <= row_number < len(X_test):
    transaction = X_test.iloc[[row_number]]

    prediction = loaded_model.predict(transaction)[0]
    probability = loaded_model.predict_proba(transaction)[0][1]

    print("\n--- Fraud Detection Result ---")
    print(f"Predicted fraud probability: {probability:.2%}")

    if prediction == 1:
        print("Result: Suspicious transaction — likely FRAUD")
    else:
        print("Result: Likely LEGITIMATE transaction")
else:
    print("Invalid row number. Please enter a valid test-set row.")
```
